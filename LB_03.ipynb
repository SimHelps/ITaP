## Задание

Напишите функцию, которая по трем переданным натуральным числам возвращает могут ли эти числа являться членами геометрической прогрессии. Если «да», то функция возвращает True, в противном случае – False.

## Решение

### Объявление функции

```python
def geom(a, b, c):
    return b * b == a * c
```

### Демонстрация использования функции

```python
def is_geom(a, b, c):
    return b * b == a * c

print(is_geom(2, 4, 8))
print(is_geom(1, 3, 9))
print(is_geom(2, 5, 10))
```
