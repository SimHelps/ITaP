## Задание

Необходимо вычислить значение выражения:
$$U = \frac {35}{a} \cdot z + z \cdot a - \frac {a + E \cdot t}{4}$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $z, a, E, t$. Переменная, которая должна быть вычислена на основе независимых: $U$.

```python
z=10
a=30.67
E=5
t=5
print("z = ", z, ", type(z) = ", type(z))
print("z = ", z, ", type(z) = ", type(z))
print("E = ", E, ", type(E) = ", type(E))
print("t = ", t, ", type(t) = ", type(t))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $U$:

```python
U = (35 / a) * z + z * a - (a + E * t) / 4
print(f"U =",U)
print('Type of U: ', type(U))
```

Основные функции, используемые в данной практической работе: `print(), type(), abs()`
