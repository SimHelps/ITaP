буквами.

## Решение

Файл pr6_1.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract6/pr6_1.txt
```

```python
f = open('pr6_1.txt')
s = f.read()
f.close()

cifri = "0123456789"
bukvy = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

count = 0

for i in range(len(s) - 3):
    s1 = s[i]
    s2 = s[i+1]
    s3 = s[i+2]
    s4 = s[i+3]

    cnt_cifr = 0
    if s1 in cifri: cnt_cifr = cnt_cifr + 1
    if s2 in cifri: cnt_cifr = cnt_cifr + 1
    if s3 in cifri: cnt_cifr = cnt_cifr + 1
    if s4 in cifri: cnt_cifr = cnt_cifr + 1

    cnt_bukv = 0
    if s1 in bukvy: cnt_bukv = cnt_bukv + 1
    if s2 in bukvy: cnt_bukv = cnt_bukv + 1
    if s3 in bukvy: cnt_bukv = cnt_bukv + 1
    if s4 in bukvy: cnt_bukv = cnt_bukv + 1

    if cnt_cifr == 2 and cnt_bukv == 2:
        count = count + 1

print(count)
```
